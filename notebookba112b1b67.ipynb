```python
import os

print("=== ALL ATTACHED DATASETS ===\n")
for root, dirs, files in os.walk('/kaggle/input'):
    level = root.replace('/kaggle/input', '').count(os.sep)
    if level <= 1:
        subdirs = [d for d in dirs if os.path.isdir(os.path.join(root, d))]
        if subdirs:
            print(f"📁 {root}")
            for d in sorted(subdirs)[:6]:
                print(f"   └── {d}")

print("\n=== SEARCHING FOR PRIMARY DATASET (Training + Testing) ===")
PRIMARY_DIR = None
for root, dirs, files in os.walk('/kaggle/input'):
    subdirs = [d for d in dirs if os.path.isdir(os.path.join(root, d))]
    has_train = any(x.lower() in ['training', 'train'] for x in subdirs)
    has_test = any(x.lower() in ['testing', 'test'] for x in subdirs)
    if has_train and has_test:
        PRIMARY_DIR = root
        print(f"✅ PRIMARY: {root}")
        print(f"   Subfolders: {subdirs}")
        break

print("\n=== SEARCHING FOR SARTAJ ===")
SARTAJ_DIR = None
for root, dirs, files in os.walk('/kaggle/input'):
    if 'sartaj' in root.lower():
        subdirs = [d for d in dirs if os.path.isdir(os.path.join(root, d))]
        SARTAJ_DIR = root
        print(f"✅ SARTAJ: {root}")
        print(f"   Subfolders: {subdirs}")
        break

if not PRIMARY_DIR:
    print("\n❌ PRIMARY DATASET NOT FOUND. Attach it via + Add Input.")
if not SARTAJ_DIR:
    print("\n❌ SARTAJ NOT FOUND.")
```
